```python
leagues = pd.DataFrame(columns=leagues_df.columns)
for div in soup.find_all('div', attrs={'class': 'gridCell Tournament Header'}):
    rows = leagues_df[leagues_df['name'].str.contains(div.find_all('a')[-1].getText())]
    leagues = pd.concat([leagues, rows], ignore_index=True)
```

```python
leagues = leagues.drop_duplicates()
leagues['liquipedia_tier'] = 'Tier 3' #TODO edit when used
```

```python
dbf.create_table_from_df(leagues, 'leagues', conn_str)
```

```python
dbf.insert_df_into_table(leagues, 'leagues', conn_str)
```

```python
cols = {
"match_id": 3703866531,
"duration": 0,
"start_time": 0,
"radiant_team_id": 0,
"radiant_name": "string",
"dire_team_id": 0,
"dire_name": "string",
"leagueid": 0,
"league_name": "string",
"series_id": 0,
"series_type": 0,
"radiant_score": 0,
"dire_score": 0,
"radiant_win": True,
"radiant": True
}
```

```python
matches = pd.DataFrame(columns=cols.keys())
for idx, row in leagues.iterrows():
    response = httpx.get(f'{api_url}/leagues/{row["leagueid"]}/matches').json()
    matches_data = pd.DataFrame(response)
    matches = pd.concat([matches, matches_data], ignore_index=True)
```

```python
matches = matches.infer_objects()
matches_dt = matches['start_time'].apply(lambda x: datetime.fromtimestamp(x))
matches.insert(2, 'start_date', matches_dt)
matches['liquipedia_tier'] = 'Tier 3' #TODO edit when used
```

```python
dbf.create_table_from_df(matches, 'matches', conn_str)
```

```python
dbf.insert_df_into_table(matches, 'matches', conn_str)
```
