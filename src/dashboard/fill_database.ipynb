```python
import pandas as pd
import numpy as np
import kagglehub

import os
import sys
sys.path.append(os.path.abspath('./src'))
from pathlib import Path

from db_functions import DotaDB
```

```python
db = DotaDB(schema='kaggle')
root_path = Path(kagglehub.dataset_download(handle='bwandowando/dota-2-pro-league-matches-2023', output_dir=f'{os.getcwd()}/data/'))
tables = {f.name for f in root_path.rglob('*.csv')}
tables.remove('cosmetics.csv')
```

```python
for table in tables:
    if table in ['Constants.Leagues.csv', 'Constants.Regions.csv']:
        df = pd.read_csv(f'{root_path}/Constants/{table}')
        try:
            df = df.drop(columns=['Unnamed: 0'])
        except:
            pass
        db_table = table[:-4].replace('.', '_')
        db.create_table_from_df(df, db_table)
        db.insert_df_into_table(df, db_table)
```

```python
data_tables = []
for table in tables: 
    if not table.startswith('Constants.'):
        data_tables.append(table)
```

```python
table_created = False
for path in root_path.glob('20*/'):
    df = pd.read_csv(f'{path}/{table}').drop(columns=['Unnamed: 0'], errors='ignore')
    if not table_created:
        db.create_table_from_df(df, 'players', add_serial_id=True)
        table_created = True
    db.insert_df_into_table(df, 'players')
data_tables.remove('players.csv')
```

```python
for table in data_tables:
    df = pd.DataFrame()
    for path in root_path.glob('20*/'):
        if '2026' in str(path):
            if table == 'draft_timings.csv':
                continue
        if df.empty:
            df = pd.read_csv(f'{path}/{table}').drop(columns=['Unnamed: 0'], errors='ignore')
        else:
            df = pd.concat([
                df,
                pd.read_csv(f'{path}/{table}').drop(columns=['Unnamed: 0'], errors='ignore')
            ])
    db.create_table_from_df(df, table[:-4], add_serial_id=True)
    db.insert_df_into_table(df, table[:-4])
```
