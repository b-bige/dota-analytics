```python
import psycopg
import os
import sys
from dotenv import load_dotenv
sys.path.append(os.path.abspath('./src'))
load_dotenv()

user = os.getenv("DB_USER")
password = os.getenv("DB_PASSWORD")
host = os.getenv("DB_HOST")
port = os.getenv("DB_PORT")
dbname = os.getenv("DB_NAME")
conn_str = f"postgresql://{user}:{password}@{host}:{port}/{dbname}"

new_host = os.getenv("NEW_DB_HOST")
new_password = os.getenv("NEW_DB_PASSWORD")
new_conn_str = f"postgresql://{user}:{new_password}@{new_host}:{port}/{dbname}"

def count_rows(cur: psycopg.Cursor):
    total = 0
    try:  
        # Get all table names
        cur.execute("""
            SELECT table_name 
            FROM information_schema.tables 
            WHERE table_schema = 'public'
        """)
        tables = [row[0] for row in cur.fetchall()]
        
        print(f"{'Table Name':<30} | {'Row Count':<10}")
        print("-" * 45)
        
        for table in tables:
            cur.execute(f'SELECT COUNT(*) FROM "{table}"')
            count = cur.fetchone()[0]
            print(f"{table:<30} | {count:<10}")
            total += count
        return total
    except Exception as e:
        print(f"Error: {e}")
with psycopg.connect(conn_str) as old_conn:
    old_cur = old_conn.cursor()
    old_total = count_rows(old_cur)
with psycopg.connect(new_conn_str) as new_conn:
    new_cur = new_conn.cursor()
    new_total = count_rows(new_cur)
if old_total == new_total:
    print(f"Total rows ({new_total}) match; migration successful")
else:
    print(f"Total rows don't match; \nOld total: {old_total} \nNew total: {new_total}")
```
